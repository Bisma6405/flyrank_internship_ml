## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Problem Framing and Business Impact


* **The Core Question**: How can we use historical traffic and performance drops to find which old pages will give the biggest traffic boost if we update them?
* **Unit of Analysis**: A single web page URL.
* **The Output**: A ranked review queue. It will list pages from highest priority to lowest priority. Each page will have an opportunity score and a reason code explaining why it needs an update.
* **The Decision**: Which old pages should get writing resources and budget instead of creating brand-new content?
* **The Action**: The content writing team will open the top-ranked pages in the queue. They will rewrite, update, or improve the text on those pages.
* **The Cost of a Wrong Call**: If the model recommends a bad page, writers will waste time and money updating content that will not gain any new traffic.
* **Why Data/ML Helps**: The website has thousands of pages. A human cannot check every page's daily history. Machine Learning can look at all the signals instantly to find hidden patterns.
